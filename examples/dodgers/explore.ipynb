```sqlite
.load ../../dist/debug/jiff0
drop table if exists schedule;
create table schedule as 
  select 
    -- '3/18/25' to  '2024-03-18'
    jiff_date_strptime('%m/%d/%y', "Game Date") as game_date,
    --"Day of Week" as day_of_week,
    "Local Time" as local_time,
    "Time - ET" as time_et,
    --"Game" as game,
    "Away Team" as away_team,
    "Home Team" as home_team,
    "Location" as location
  from "mlb_schedule_2025_for_reddit - mlb_schedule_2025.csv";
```

```sqlite
select *
  --jiff_time_strptime('%-I:%M %P', local_time)
from schedule
where (home_team = 'Dodgers' or away_team = 'Dodgers')
  and game_date > date('now');
```
